```python
from langchain_community.document_loaders import WebBaseLoader
```

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
loader = WebBaseLoader(web_paths=["https://www.educosys.com/course/genai"])

docs = loader.load()
print(docs)
```

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(chunk_size = 1000, chunk_overlap = 200)
splits = text_splitter.split_documents(docs)
```

```python
for i, split in enumerate(splits):
    print(f"Split {i+1}:")
    print(split.page_content[:200])  # Print first 200 characters of each split
    print("-----")
```

```python
from  langchain_huggingface import HuggingFaceEmbeddings
embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
from langchain_community.vectorstores import Chroma
vectorstore = Chroma.from_documents(splits, embeddings, persist_directory="chroma_db")
```

```python
print(vectorstore._collection.count())
```

```python
print(vectorstore._collection.get())
```

```python
print("\n Collection 1: ", vectorstore._collection.get(ids=["5fa0c1b6-dc20-4443-ad96-d8db099afbac"], include=[ "metadatas",  "documents"]))
```

```python
retriver = vectorstore.as_retriever()
```

```python
# Example query
query = "What is the course about?"
results = retriver.get_relevant_documents(query)
for i, result in enumerate(results):
    print(f"Result {i+1}:")
    print(result.page_content[:200])  # Print first 200 characters of each result
    print("-----")
```

```python
#rag chain
from langchain.chains import RetrievalQA
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="llama-3.1-8b-instant",
    temperature=0.0,
    max_retries=2,
    # other params...
)

rag_chain = RetrievalQA.from_chain_type(
    llm=llm,
    chain_type="stuff",
    retriever=retriver,
    return_source_documents=True
)

query = "What is the course about?"
result = rag_chain({"query": query})
print(f"Query: {query}")
print(f"Answer: {result['result']}")
```
